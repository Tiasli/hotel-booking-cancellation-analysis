```python
from scipy import stats

#ANOVA
 
df_between = 11
df_within = 348 #360-12 (degrees of freedom is substracted)
alpha = 0.05

f_crit = stats.f.ppf(1 - alpha, df_between, df_within)
print(f"F crit: {f_crit}")


#F value

df_clean = df.dropna(subset=['is_canceled', 'arrival_date_month']).copy()
df_clean['arrival_date_month'] = df_clean['arrival_date_month'].str.strip() #Values were turned into string and got cleaned from empty spaces 

grouped_data = df_clean.groupby(['arrival_date_month', 'arrival_date_day_of_month'])['is_canceled'].mean().reset_index()

months_ordered = ['January', 'February', 'March', 'April', 'May', 'June', 
                  'July', 'August', 'September', 'October', 'November', 'December']

anova_groups = []
for m in months_ordered:
    group = grouped_data[grouped_data['arrival_date_month'] == m]['is_canceled']
    if len(group) > 0: #only if there is data        
        anova_groups.append(group)

f_stat, p_val = stats.f_oneway(*anova_groups)
print(f"F value: {f_stat}")
print(f"p value: {p_val}")
```

```python

from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey = pairwise_tukeyhsd(endog=grouped_data['is_canceled'], 
                          groups=grouped_data['arrival_date_month'], 
                          alpha=0.05)
print(tukey)
```

```python
import matplotlib.pyplot as plt

# Tukey result
fig = tukey.plot_simultaneous(comparison_name='January', ylabel='Months', xlabel='Mean Difference')
plt.title('Tukey Test - January')
plt.show()
```

```python
# Sadece iptal edilenleri al
cancelled_df = df[df['is_canceled'] == 1].copy()

# Lead Time'ları say ve sırala
pareto_data = cancelled_df['lead_time'].value_index().sort_values(ascending=False)

# Kümülatif toplam ve yüzde
cumulative_sum = pareto_data.cumsum()
cumulative_perc = 100 * cumulative_sum / cumulative_sum.max()

# %80'in nerede bittiğini bul
pareto_threshold = cumulative_perc[cumulative_perc <= 80].index[-1]
print(f"İptallerin %80'i, {pareto_threshold} gün lead time'a kadar olan gruptan geliyor.")
```

```python
# 1. Adım: Tarih sütununu gerçek bir tarih objesine çevir
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])

# 2. Adım: Veriyi tarihe göre kronolojik olarak sırala
df.sort_values('reservation_status_date', inplace=True)

# 3. Adım: "City Hotel" verisini ayır
city_hotel = df[df['hotel'] == 'City Hotel']
resort_hotel = df[df['hotel'] == 'Resort Hotel']

# 4. Adım (KRİTİK): Günlük bazda veriyi grupla (Örn: Günlük ortalama fiyat veya toplam iptal)
# Eğer her satırı çizmeye çalışırsan spagetti olur. Gruplamak şart.
city_hotel = city_hotel.set_index('reservation_status_date')[['is_canceled']].resample('ME').mean()
resort_hotel = resort_hotel.set_index('reservation_status_date')[['is_canceled']].resample('ME').mean()


# 5. Adım: Şimdi çizdir
plt.figure(figsize=(15, 6))
plt.plot(resort_hotel.index, resort_hotel['is_canceled'], label = "Resort Hotel")
plt.plot(city_hotel.index, city_hotel['is_canceled'], label = "City Hotel")
plt.title("Cancelation Rate in City and Resort Hotel Over Time", fontsize = 30)
plt.legend(fontsize = 10)
plt.show()


```

```python
active_cancels = df[df['reservation_status'] == 'Canceled'].copy()

# 2. Ensure the status date is in datetime format
active_cancels['reservation_status_date'] = pd.to_datetime(active_cancels['reservation_status_date'])

# 3. Group by the date of the action
# We count how many people 'hit the button' each month
cancellation_trend = active_cancels.groupby('reservation_status_date').size().resample('ME').sum()

# 4. Plot the "Decision Timeline"
plt.figure(figsize=(15, 6))
plt.plot(cancellation_trend.index, cancellation_trend.values, color='#d63031', linewidth=2)

plt.title("Cancellation Timeline", fontsize=16)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Number of Cancellations", fontsize=12)
plt.grid(True, alpha=0.2)
plt.show()
```
